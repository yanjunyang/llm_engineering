```python
# imports

import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, without streaming
openai = OpenAI()

def get_gpt_answer_complete_response(question):
    response = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=[{"role": "user", "content": question}]
    )
    result =response.choices[0].message.content
    display(Markdown(result))
```

```python
get_gpt_answer_complete_response(question)
```

```python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI()

def get_gpt_answer(question):
    stream = openai.chat.completions.create(
        model=MODEL_GPT,
        messages=[{"role": "user", "content": question}],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        update_display(Markdown(response), display_id=display_handle.display_id)
    
```

```python
result = get_gpt_answer(question)
```

```python
# Get Llama 3.2 to answer
import requests
# Check if ollma is running
requests.get("http://localhost:11434").content
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python
def get_ollama_answer(question):
    stream = ollama.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[{"role": "user", "content": question}],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        update_display(Markdown(response), display_id=display_handle.display_id)
    
```

```python
get_ollama_answer(question)
```
