## And now for some fun - an adversarial conversation between Chatbots..

You're already familar with prompts being organized into lists like:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

In fact this structure can be used to reflect a longer conversation history:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

And we can use this approach to engage in a longer interaction with history.

````python
# Let's make a conversation between GPT-4.1-mini and Claude-haiku-4.5
# We're using cheap versions of models so the costs will be minimal

gpt_model = "gpt-4.1-mini"
claude_model = "claude-haiku-4-5"

gpt_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

claude_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
# Let's make a conversation between GPT-4.1-mini and llma 3.2
# We're using cheap versions of models so the costs will be minimal

gpt_model = "gpt-4.1-mini"
llama_model = "llama3.2"

gpt_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

llama_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gpt_messages = ["Hi there"]
llama_messages = ["Hi"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, llama in zip(gpt_messages, llama_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": llama})
    response = openai.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_gpt()
````

````python
def call_claude():
    messages = [{"role": "system", "content": claude_system}]
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = anthropic.chat.completions.create(model=claude_model, messages=messages)
    return response.choices[0].message.content
````

````python
def call_llama():
    messages = [{"role": "system", "content": llama_system}]
    for gpt, llama_message in zip(gpt_messages, llama_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": llama_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = ollama.chat.completions.create(model=llama_model, messages=messages)
    return response.choices[0].message.content
````

## Additional exercise

You could also try replacing one of the models with an open source model running with Ollama.

````python

# 3 models, each model has its own system prompt
# There is only one list to save all chats history
# Everytime a response generates, append to the chat_hist

import os
import requests # test ollma running status
from dotenv import load_dotenv 
from openai import OpenAI
from IPython.display import Markdown, display

ollama_url = "http://localhost:11434/v1"
ollama_model = "llama3.2"
gpt_model = "gpt-4.1-mini"
````

````python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
````

````python
requests.get("http://localhost:11434/").content
````

````python
openai = OpenAI()
ollama1 = OpenAI(api_key="ollama", base_url=ollama_url)
ollama2 = OpenAI(api_key="ollama", base_url=ollama_url)
````

````python
alex_system = "You are Alex, a chatbot who is very argumentative; you disagree with anything in the conversation and you challenge everything, in a snarky way."
blake_system = "You are Blake, a chatbot who supports LA Lakers. You will support the Lakers no matter what, even if they are losing. And you will become very agrresive if anyone says anything bad about the Lakers."
charlie_system = "You are Charlie, a chatbot who supports LA Clippers. You will support the Clippers no matter what, even if they are losing. But you are a peaceful guy evenif anyone says anything bad about the Clippers. And you don't want to fight with anyone."

alex_prompt_template = """
You are Alex, in conversation with Blake and Charlie.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Alex.
"""

blake_prompt_template = """
You are Blake, in conversation with Alex and Charlie.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Blake.
"""

charlie_prompt_template = """
You are Charlie, in conversation with Alex and Blake.
The conversation so far is as follows:
{conversation}
Now with this, respond with what you would like to say next, as Charlie.
"""
````

````python
conversation = []  # one shared transcript
conversation.append("Alex: Hi everyone")
conversation.append("Blake: Lakers forever!")
conversation.append("Charlie: Go Clippers, peacefully. But we beat Lakers today. What a good day.")

conversation_text = "\n".join(conversation)
print(conversation_text)
````

````python
def call_alex():
    response = openai.chat.completions.create(
        model=gpt_model,
        messages=[{"role": "system", "content": alex_system}, 
                {"role": "user", "content": alex_prompt_template.format(conversation=conversation_text)}]
    )
    return response.choices[0].message.content

def call_blake():
    response = ollama1.chat.completions.create(
        model=ollama_model,
        messages=[{"role": "system", "content": blake_system}, 
                {"role": "user", "content": blake_prompt_template.format(conversation=conversation_text)}]
    )
    return response.choices[0].message.content

def call_charlie():
    response = ollama2.chat.completions.create(
        model="qwen3.8:27b-mlx",
        messages=[{"role": "system", "content": charlie_system}, 
                {"role": "user", "content": charlie_prompt_template.format(conversation=conversation_text)}]
    )
    return response.choices[0].message.content
````

````python
def conversation_update(conversation, response, character):
    conversation.append(f"{character}: {response}")
    conversation_text = "\n".join(conversation)
    return conversation_text
````

````python
response = call_alex()
conversation_text = conversation_update(conversation, response, "Alex")
````

````python
print(conversation_text)
````

````python
response = call_blake()
display(Markdown(response))
````

````python
print(f"Initial conversation:\n{conversation_text}")

for i in range(5):
    response = call_alex()
    conversation_text = conversation_update(conversation, response, "Alex")
    display(Markdown(f"### Alex:\n{response}\n"))

    response = call_blake()
    conversation_text = conversation_update(conversation, response, "Blake")
    display(Markdown(f"### Blake:\n{response}\n"))

    response = call_charlie()
    conversation_text = conversation_update(conversation, response, "Charlie")
    display(Markdown(f"### Charlie:\n{response}\n"))
````

## Part 2: Dynamic turn-taking with interruptions (prototype)

This extends Part 1 toward more realistic group chat behavior using an **event loop** instead of fixed rounds.

### Idea
1. **Start step** — `floor_holder` is whoever currently has the floor.
2. **Speak chunk** — floor holder generates one short chunk (1–2 sentences).
3. **Interrupt check** — each other speaker answers: cut in now? (YES / NO).
4. **If someone interrupts:**
   - Save partial line as `… [cut off]` in the transcript
   - Interrupter speaks
   - `floor_holder` = interrupter
   - Go to the next step
5. **If no interrupt:**
   - Commit the full chunk to the transcript
   - **45% chance:** same speaker keeps the floor → next step
   - **Otherwise:** pass floor to a **random** other speaker → next step
6. Repeat for `steps` speaking events (default: 12).

### Configuration

- `speakers` dict — maps each name to `{client, model, system}`
- `conversation` — same shared transcript list as Part 1
- `floor_holder` — who currently has the floor
- `steps` — number of speaking events (default 12), not “rounds”

### Core functions

| Function | Purpose |
|----------|---------|
| `transcript()` | Join `conversation` into a single string |
| `call_speaker(name, instruction)` | Generate 1–2 sentences for a speaker using `system` + `user` (full transcript + instruction). `max_tokens=80` keeps chunks short. |
| `wants_to_interrupt(listener, floor_holder, partial)` | Cheap YES/NO check: should `listener` cut in while `floor_holder` has only said `partial` so far? `max_tokens=3`, low temperature. |
| `pick_interrupter(floor_holder, partial)` | Ask all non-speakers (shuffled order); return a random one who said YES, or `None` |
| `show(name, text)` | Display markdown in the notebook |

### Per-step flow

1. **Chunk** — `floor_holder` generates a partial utterance via `call_speaker()`
2. **Interrupt check** — `pick_interrupter()` polls the other speakers
3. **If interrupted:**
   - Append `{speaker}: {partial} … [cut off]` to transcript
   - Interrupter speaks with instruction *"You interrupted X mid-sentence. Jump in now."*
   - `floor_holder = interrupter`
4. **If not interrupted:**
   - Append full chunk to transcript
   - With probability 0.45, same person keeps the floor (`continue`)
   - Otherwise, `floor_holder` passes to a **random** other speaker

### What this simulates vs. real speech

| Real behavior | How we approximate it |
|---------------|------------------------|
| Not fixed order | `floor_holder` changes dynamically |
| Cut in mid-sentence | Partial chunk + `[cut off]` marker |
| Everyone listening | Each listener gets the partial text before deciding |
| Not always same interrupter | `random.shuffle(candidates)` |

Still **turn-based text simulation** — not true parallel audio/streaming. Overlap is implied by short chunks and interrupt decisions.

## Possible next improvements (not implemented yet)

2. **Urgency score (0–10)** instead of YES/NO for finer interrupt decisions
3. **Interrupt cooldown** — prevent the same person interrupting every turn
4. **Summarize old transcript** when context gets long

````python
import random
from IPython.display import Markdown, display

alex_model = "gpt-4.1-mini"
blake_model = "llama3.2"
charlie_model = "llama3.2"  # or llama3.2

ollama = OpenAI(api_key="ollama", base_url=ollama_url)
openai = OpenAI()

speakers = {
    "Alex":   {"client": openai, "model": alex_model, "system": "You are Alex: snarky, argumentative. 1-2 sentences max."},
    "Blake":  {"client": ollama, "model": blake_model, "system": "You are Blake: calm, polite. 1-2 sentences max."},
    "Charlie":{"client": ollama, "model": charlie_model, "system": "You are Charlie: curious, upbeat. 1-2 sentences max."},
}

conversation = [
    "Alex: So pineapple on pizza is obviously correct.",
    "Blake: I hear you, but maybe we can keep this friendly?",
]

def transcript() -> str:
    return "\n".join(conversation)

def call_speaker(name: str, instruction: str) -> str:
    cfg = speakers[name]
    messages = [
        {"role": "system", "content": cfg["system"]},
        {"role": "user", "content": f"""
        You are {name}, talking with the others.

        Conversation so far:
        {transcript()}

        Instruction:
        {instruction}

        Reply with ONLY what {name} says next (no name prefix).
        Keep it to 1-2 short sentences.
        """},
    ]
    r = cfg["client"].chat.completions.create(
        model=cfg["model"],
        messages=messages,
        max_tokens=80,
    )
    # .strip() removes leading/trailing whitespace, especially newlines and spaces.
    return r.choices[0].message.content.strip()

def wants_to_interrupt(listener: str, floor_holder: str, partial: str) -> bool:
    """Cheap yes/no check: should this person cut in right now?"""
    cfg = speakers[listener]
    messages = [
        {"role": "system", "content": cfg["system"]},
        {"role": "user", "content": f"""
        You are {listener}, listening to a group chat.

        Conversation so far:
        {transcript()}

        Right now {floor_holder} is speaking and has only said so far:
        "{partial}"

        Would you cut in right now?
        Answer with exactly one word: YES or NO.
        Say YES if you have something relevant to add or disagree with.
        """},
    ]
    r = cfg["client"].chat.completions.create(
        model=cfg["model"],
        messages=messages,
        max_tokens=3,
        temperature=0.2,
    )
    return r.choices[0].message.content.strip().upper().startswith("Y")

def pick_interrupter(floor_holder: str, partial: str) -> str | None:
    candidates = [n for n in speakers if n != floor_holder]
    random.shuffle(candidates)  # not always same person jumping in

    yes_speakers = [n for n in candidates if wants_to_interrupt(n, floor_holder, partial)]
    return random.choice(yes_speakers) if yes_speakers else None

def show(name: str, text: str):
    display(Markdown(f"### {name}\n{text}\n"))

# --- Simulation ---
floor_holder = "Alex"
steps = 12  # total speaking events (not fixed rounds)

for step in range(steps):
    display(Markdown(f"---\n## Step {step + 1}"))

    # 1) Current speaker starts / continues a turn in chunks
    partial = call_speaker(
        floor_holder,
        f"You currently have the floor. Say the next short chunk of what you were saying."
    )
    show(floor_holder, partial + " …")

    # 2) Maybe someone interrupts mid-thought
    interrupter = pick_interrupter(floor_holder, partial)

    if interrupter:
        conversation.append(f"{floor_holder}: {partial} … [cut off]")
        show(floor_holder, f"partial Commited: {partial} … **[cut off]**")
        
        display(Markdown(f"### Interrupter\n{interrupter}\n"))
        interrupt_line = call_speaker(
            interrupter,
            f"You interrupted {floor_holder} mid-sentence. Jump in now."
        )
        conversation.append(f"{interrupter}: {interrupt_line}")
        show(interrupter, interrupt_line)

        floor_holder = interrupter
        continue

    # 3) No interrupt: commit chunk, maybe continue same turn or hand off floor
    conversation.append(f"{floor_holder}: {partial}")
    show(floor_holder, f"No interrupt Commited: {partial}")

    keep_talking_score = random.random()
    print(f"Keep talking score: {keep_talking_score}")
    if keep_talking_score < 0.45:
        # same person keeps talking
        continue

    # hand floor to someone else (not strict rotation)
    others = [n for n in speakers if n != floor_holder]
    floor_holder = random.choice(others)
````

## Part 3: Orchestrator-driven turn-taking

Part 2 polls each listener with YES/NO and randomly passes the floor. Part 3 replaces that with **one `gpt-4.1-mini` orchestrator call per step** that decides who speaks next and whether they interrupt.

### Problem Part 3 solves

| Part 2 limitation | Part 3 fix |
|-------------------|------------|
| 2 interrupt API calls every step (Blake + Charlie) | 1 orchestrator call per step |
| Random floor handoff (`random.choice`) | Model picks the next speaker intentionally |
| YES/NO only — no explanation | `reason` field explains each decision |
| Interrupt logic spread across 3 functions | Centralized in `orchestrate()` |

### Models & clients

| Role | Client | Model |
|------|--------|-------|
| Alex (speaker) | `openai` | `gpt-4.1-mini` |
| Blake (speaker) | `ollama1` | `llama3.2` |
| Charlie (speaker) | `ollama2` | `llama3.2` |
| Orchestrator (moderator) | `openai` | `gpt-4.1-mini` |

All speakers still use the **1 system + 1 user** pattern with the full labeled transcript. The orchestrator also sees the full transcript plus the current partial chunk.

### Shared state

| Variable | Purpose |
|----------|---------|
| `conversation` | List of `"Name: message"` strings — single shared transcript |
| `floor_holder` | Who currently has the floor |
| `last_interruptor` | Who interrupted last — orchestrator is told not to let them interrupt again immediately |
| `steps` | Number of speaking events to run (default: 5 in code) |

### Orchestrator JSON response

Each step, `orchestrate()` returns:

```json
{
  "speaker": "Blake",
  "interrupt": true,
  "reason": "Blake wants to jump in before Alex finishes his point"
}

````python
import json
from IPython.display import Markdown, display
ORCHESTRATOR_MODEL = "gpt-4.1-mini"

openai = OpenAI()
ollama1 = OpenAI(api_key="ollama", base_url=ollama_url)
ollama2 = OpenAI(api_key="ollama", base_url=ollama_url)

speakers = {
    "Alex": {
        "client": openai,
        "model": "gpt-4.1-mini",
        "system": "You are Alex: snarky, argumentative. 1-2 sentences max.",
    },
    "Blake": {
        "client": ollama1,
        "model": "llama3.2",
        "system": "You are Blake: calm Lakers fan, polite but passionate. 1-2 sentences max.",
    },
    "Charlie": {
        "client": ollama2,
        "model": "llama3.2",
        "system": "You are Charlie: peaceful Clippers fan, curious. 1-2 sentences max.",
    },
}

conversation = [
    "Alex: So pineapple on pizza is obviously correct.",
    "Blake: I hear you, but maybe we can keep this friendly?",
]

last_interruptor = None

def transcript() -> str:
    return "\n".join(conversation)

def call_speaker(name: str, instruction: str) -> str:
    cfg = speakers[name]
    messages = [
                {"role": "system", "content": cfg["system"]},
                {"role": "user", "content": f"""
                You are {name}, in a group chat with the others.
                Conversation so far:
                {transcript()}
                Instruction:
                {instruction}
                Reply with ONLY what {name} says next (no name prefix).
                Keep it to 1-2 short sentences.
                """},
            ]
    r = cfg["client"].chat.completions.create(
        model=cfg["model"],
        messages=messages,
        max_tokens=80,
    )
    return r.choices[0].message.content.strip()
    
def orchestrate(floor_holder: str, partial: str, last_interruptor: str | None) -> dict:
    system = """You are the moderator of a 3-person group chat between Alex, Blake, and Charlie.
        Decide who should speak next and whether they are interrupting the current speaker.
        Rules:
        - speaker must be exactly one of: Alex, Blake, Charlie
        - interrupt=true only if the chosen speaker cuts in BEFORE floor_holder finishes their thought
        - interrupt=false if floor_holder's chunk should be committed and speaker continues or responds normally
        - If partial chunk is short or incomplete, interrupt is more likely
        - Vary turn-taking naturally — do not always pick the same speaker
        - If last_interruptor is set, that person should NOT interrupt again
        - Return ONLY valid JSON"""
    
    user = f"""
        Conversation so far:
        {transcript()}
        Current floor_holder: {floor_holder}
        Partial chunk (may be incomplete): "{partial}"
        Last interrupter: {last_interruptor or "none"}
        Return JSON with exactly these fields:
        - speaker: string (Alex, Blake, or Charlie)
        - interrupt: boolean
        - reason: string (one short sentence)
        """
    r = openai.chat.completions.create(
        model=ORCHESTRATOR_MODEL,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": user},
        ],
        response_format={"type": "json_object"},
        temperature=0.4,
        max_tokens=120,
    )
    return json.loads(r.choices[0].message.content)

def show(name: str, text: str):
    display(Markdown(f"### {name}\n{text}\n"))

floor_holder = "Alex"
steps = 5

for step in range(1, steps + 1):
    display(Markdown(f"---\n## Step {step}"))
    partial = call_speaker(
        floor_holder,
        "You have the floor. Say the next short chunk of what you want to say.",
    )
    show(floor_holder, partial + " …")
    decision = orchestrate(floor_holder, partial, last_interruptor)
    speaker = decision["speaker"]
    interrupt = decision["interrupt"]
    reason = decision["reason"]
    display(Markdown(
        f"**Orchestrator:** `{speaker}` | interrupt=`{interrupt}` | {reason}"
    ))
    if interrupt and speaker != floor_holder:
        conversation.append(f"{floor_holder}: {partial} … [cut off]")
        show(floor_holder, f"{partial} … **[cut off]**")
        line = call_speaker(
            speaker,
            f"You interrupted {floor_holder} mid-sentence. Reason: {reason}. Jump in now.",
        )
        conversation.append(f"{speaker}: {line}")
        show(speaker, line)
        floor_holder = speaker
        last_interruptor = speaker
        continue
    conversation.append(f"{floor_holder}: {partial}")
    show(floor_holder, partial)
    if speaker != floor_holder:
        line = call_speaker(
            speaker,
            f"Respond now. Reason you speak: {reason}",
        )
        conversation.append(f"{speaker}: {line}")
        show(speaker, line)
        floor_holder = speaker
    else:
        floor_holder = speaker
print("\n--- Final transcript ---")
print(transcript())
````
